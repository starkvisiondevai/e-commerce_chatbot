```python
import pandas as pd 
from langchain_core.documents import Document 

def dataConverter():
    products_data = pd.read_csv("../dataset/flipkart_product_reviews.csv")

    data = products_data[['product_title', 'review']]

    product_docs = []

    for index, row in data.iterrows():
        metadata = {"product_name" : row['product_title']}
        doc = Document(page_content=row['review'], metadata=metadata)
        product_docs.append(doc)

    return product_docs
```

```python
product_docs = dataConverter()
#product_docs[0].metadata.values()
for i in range(len(product_docs)):
    print(product_docs[i].metadata['product_name'][0])
```
